```python
numbers = pd.Series(
    [7, 9, 10, 11],  # : values
    index=['ronaldo', 'benzema', 'james', 'bale'], # 각각의 index가 누구것이냐
    name='JNo.',
    dtype='int64'
)

print(f'values = [numbers.values]')
print(numbers)

print(type(numbers.values)) #numpy배열로 변환해준다
print(f'index = {numbers.index.to_list()}')
print(f'dtype = {numbers.dtype}')
print(f'name = {numbers.name}')
print(f'size = {numbers.size}')
print(f'shape = {numbers.shape}')
print(f'ndim = {numbers.ndim}')
print(f'empty = {numbers.empty}')

if not numbers.empty:
   pass #나중에 코드를 짜겠다는 표시(덜 짜도 오류x)
```
