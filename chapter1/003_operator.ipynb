### 연산자(operator): 특정 계산이나 처리를 수행하도록 지시하는 기호 또는 키워드

 1 + 1

* 피연산자(Oeperand): 연산자에 의해서 처리되는 대상이 되는 값이나 변수

* 연산자의 구분: 연산자 자신이 처리하는 피연산자의 개수에 따라 구분 가능

1. 단항연산자(Unary Operator)
    단항 연산자는 하나의 피연산자에 대해서 연산을 수행하는 연산자들을 의미한다.

2. 이항연산자(Binary Operator)
    두 개의 피연산자에 대해서 연산을 수행하는 연산자

3. 다항연산자(Ternary Operator), 삼항연산자
    피연산자를 세 개 혹은 그 이상을 요구하는 연산자


#### 기능에 따른 구분

##### 산술 연산자(Arithmetic Operator)
* 수학적 계산을 수행
* 한 개 혹은 두 개의 숫자의 수학적인 작업 처리

```python
# 덧셈 연산자

a, b, c, d, e = 7, 77, 0.77, True, False

print(a + b + c + d + e)
print(a + b + c)
print(a + b + c + d)

a, b, c, d = "pizza", 10, 0.3 , True
b + c + d

hello_str = "Hello, "
world_str = "World!"
result = hello_str + world_str
print(result)

list1 = [1,2,3]
list2 = [4,5,6]
print(list1 + list2)

print(list1 * 3)
```

```python
# 뺄셈 연산자

a, b, c = 7, 77 ,0.77

integer_result = b - a #or a - b
print(integer_result)
```

```python
# 곱셈 연산자
print(a * b)

print("피자" * 10)

## 제곱
print(a ** b)
```

```python
# 나눗셈 연산자
print(10 / 4) #정수 연산
print(10 // 4) #몫만 나옴
print(10 % 4) #나머지 계산(moduler 연산자)
```

```python
# 비교 연산자

print(77 > 7)
print(77 < 7)
print(77 >= 7) #기호 순서 헷갈림
print(77 <= 7) 
print(77 == 7)
print(77 == 77.0)
print(77 != 7)

x = 5
x > 1
x < 10

1 < x and x < 10 

## 연쇄비교
1 < x < 10 
```

#### 논리 연산자(Logical Opreator): 논리값(bool)을 다룸, 여러 조건을 조합하여 복합조건을 만들때 사용 

```python
#and: 둘 다 만족

print(True and True)
print(True and False)
print((1 > 0) and (1 > -1))
```

```python
#or: 둘 중 하나

print(True or False)
print(False or True)

print(False or False)
```

```python
#not: 부정연산자(~ 가 아니라면, - 일텐데)

print(not True)
```
