### 자료형(빅데이터분석가 딸 때 중요)

> 숫자형은 숫자 넣는 곳에, 자료형(data types)은 자료 넣는 곳에

* 숫자 자료형(Numeric)
1. 정수 ~: 4bytes 만큼 가져가야함
2. 실수 ~: 부동(둥둥 떠다닌다) 소수점
* ~~문자 자료형(Text): 글자 여러개 놓기~~
* 시퀀스 자료형(Sequence): 자료 구조
* 논리 자료형(Boolean): 
* 바이트 자료형(Byte)
* None 자료형: '없다'라는 것을 의미 -> 없는게 값.(정말 아무것도 없다)


```python
### 정수 자료형
#### int(integer)

age = 25
temperature = -19999
zero = 0

big_number = 77777777777777777777777777777777777
print(big_number - 5)

seoul = 1_000_000_000
print(big_number - seoul)


#### 2진수 = binary
binary = 0b1010
print(binary)

octal = 0o017
print(octal)

hexadecimal = 0xFF
print(hexadecimal)
```

```python
### 실수(RealNumber) 자료형
#### 부동 소수점(Floating-point) 내부적으로는 IEE 754 계산을 수행

pi = 3.141592
temperature = -21.8764

some_value = .125

ten_f = 10.0
##### 실수 자료형
ten_i = 10
##### 정수 자료형
print(pi - ten_f)
print(pi - ten_i)

10.1e10

print(0.2 + 0.1)

# import: 도구 가져오기
import math
result = 0.1 +0.2
print(math.isclose(result, 0.3))

from decimal import Decimal
result = Decimal('0.1') + Decimal('0.2')
print(result)
```

```python
### 시퀀스 자료형

#### 배열(list)
numbers = [1,2,3,4]

##### *index: 양수는 0부터, 음수는 -1부터
print(numbers)
result = numbers[0]
result2 = numbers[-1]
print(result)
print(result2)

my_ary = [1, "1", True]

nested = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
#### 만약 2를 꺼내고 싶다...
nested[0][1]

first_result = nested[0]
result = first_result[1]
# = result = nested[0][1]

result = nested[0][1]
print(result)

foods = ["과일", "피자", "국밥", "치킨", "샐러드"]

foods[0:4:2]
print(foods[0:4:2])
foods[2] # 국밥
print(foods[2])
foods[-4] # 피자
print(foods[-4])
# index가 자료를 넘어가면 오류

foods[2:4] # 국밥, 치킨, 샐러드 x => 마지막 하나 빼고 가져옴
print(foods[2:4])

print(foods[0:4:2])
print(foods[0:4:3])

print(foods[::-1]) # 역순

foods.append('김밥') #추가
print(foods)

foods.remove('김밥') #제외
print(foods)
```

#### 데이터에서 가장 작은 단위: bit
- 1 byte = 8 bit
##### 1. 데이터 용량 단위 비교표 (2진수 기준)

| 단위명 | 기호 | Byte 환산 | Bit 환산 | 이전 단위 기준 | 대략적인 크기 체감 |
| :--- | :---: | :--- | :--- | :--- | :--- |
| **Bit** | b | 0.125 B | 1 bit | - | 0 또는 1 (최소 단위) |
| **Byte** | B | 1 B | 8 bit | 8 bits | 영문자/숫자 1글자 |
| **Kilobyte** | KB | 1,024 B | 8,192 bit | 1,024 B ($2^{10}$) | 짧은 텍스트 메모 (~1,000자) |
| **Megabyte** | MB | 1,048,576 B | 8,388,608 bit | 1,024 KB ($2^{20}$) | MP3 음악 1곡 (~3~5MB) |
| **Gigabyte** | GB | 1,073,741,824 B | 약 8.59억 bit | 1,024 MB ($2^{30}$) | FHD 영화 1편 (~2~4GB) |
| **Terabyte** | TB | 약 1.10조 B | 약 8.80조 bit | 1,024 GB ($2^{40}$) | 스마트폰 사진 약 25만 장 |
| **Petabyte** | PB | 약 1.13천조 B | 약 9.01천조 bit | 1,024 TB ($2^{50}$) | 4K 고화질 영상 약 3년 연속 재생 분량 |
| **Exabyte** | EB | 약 1.15백경 B | 약 9.22백경 bit | 1,024 PB ($2^{60}$) | 전 세계 데이터 트래픽 측정 단위 |
