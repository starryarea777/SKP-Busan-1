### 조건문(Conditional Statement)
: 조건문은 주어진 조건에 따라 프로그램이 서로 다른 코드 블록을 실행하도록 분기시키는 제어구조

#### if

```python
"""
기본 if문 구조
if 값이나 조건식: 명령문 
"""

temperature = 35

if temperature > 30:
    print("엄청 더운 날씨입니다.")
    print("Please drink water!")

print("현재 기온: ", temperature)

score = 77

if score >= 60:
    print('합격입니다!')

lunch = ['피자', '햄버거', '국밥', '돈까쓰']

# 맴버십 연산자: in

if '햄버거' in lunch:
    print('햄버거를 먹어보자.')

food = None

if food is None:     #동일성: 진짜냐 아니냐 #동등성: 같냐 안같냐
    print("음식이 없습니다.")
print('Go and eat out!')
```

#### if ~ elif - else (평가내용: 참인지 거짓인지)
: else if와 같이 if절 평가내용이 거짓일 경우 다른 조건을 추가하고자 할 때 사용

```python
score = 77

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70: 
    grade = "C"
elif score >= 60:
    grade = "D"
else: grade = "F"

print(grade)  #첫번째 elif에서 평가하면 평가가 끝난다.
```

#### while
: 주어진 조건식이 참인 동안에 코드블럭을 반복해서 실행

* while문 구조
'''python
while 조건식 or 값:
반복할 명령문
'''

```python
am_i_hungry = True

while am_i_hungry:
    print("Eat something")
```

```python
count = 7   #초기식

while count <= 77:     #평가절
    print(f'{count}입니다!')
    count = count + 7    #증감식

print('LUCKY!')
```

```python
n = 7

while n > 0:
    print(n, end=" ")
    n = n - 1

print('0!')
```

#### for문(python)

* for: 꺼내온 요소를 담을 변수

```python
for number in [ 1, 2, 3, 4 ]:
    print(2)

for number in range(8):
    print(number + 0)
```

### 분기문(Branching Statement)
: 분기문은 명령의 흐름을 제어하여 특정 코드블록으로 이동하거나 반복문을 종료하는 등 이전의 제어문들과 마찬가지로 명령의 흐름의 분기를 생성할 수 있는 문법

#### break
: 현재 실행중인 루프를 즉시 탈출하는 키워드

```python
for x in range(10):
    if x == 5:
        print(f'{x}에서 break가 발동!')
        break
    print(f'현재 숫자 = {x}')
```

#### continue
: 현재 반복의 나머지 코드를 건너 뛰고 다음 반복으로 즉시 이동하는 키워드

```python
for x in range(1, 7):
    print(f'이번 손님 = {x}')
    if x % 2 == 0:
        continue   #반복 순서를 건너 뛴다
    print(x)
```
