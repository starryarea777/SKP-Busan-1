### 함수(function)
: 함수는 특정한 목적을 위하여 작성된 코드의 묶음에 이름을 붙여서 재사용할 수 있도록 ***정의한 것***

ex. 'print()'

#### 함수의 구조

* python

def 함수 이름(변수1, 변수2, ...):  
    
        명령문


**(변수1, 변수2, ...) => 매게변수**


```python
print(5 + 10)
print(10 + 3)
print(10 +10)
print(1 + 7)
print(7 + 77)
...
```

```python
# 덧셈

def sum(a, b):
    print(a + b)

sum(77, 700)  #인수/인자

sum(a=7, b=70)
```

```python
def greeting():
    print('ASAP, ROCKY')
```

```python
greeting()
```

```python
# 곱셈

def multiply(a, b):
    result = a * b
    return result

multiply(7, 77)
print()
```

```python
def multiply(a, b):
    return a * b

result = multiply(77, 77)
print(result)
```

```python
def func1() -> None:
    print()

def func2(a: int, b: float) -> None:
    print()

def func3(a: int, b: float) -> int:
    return a + b

def func4(a: int, b: int) -> int:
    '''
    두 정수를 합산하는 멋진 함수
    '''
    return a + b
```
